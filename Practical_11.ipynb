```python
import time

# Graph represented using an adjacency matrix
# INF represents no direct connection
INF = float('inf')

graph = [
    [0,   5,   INF, 10],
    [INF, 0,   3,   INF],
    [INF, INF, 0,   1],
    [INF, INF, INF, 0]
]

V = len(graph)

# Start execution time
start_time = time.perf_counter()

# Copy graph to distance matrix
dist = [row[:] for row in graph]

# Floyd-Warshall Algorithm
for k in range(V):
    for i in range(V):
        for j in range(V):
            if dist[i][k] != INF and dist[k][j] != INF:
                dist[i][j] = min(
                    dist[i][j],
                    dist[i][k] + dist[k][j]
                )

# End execution time
end_time = time.perf_counter()

# Display shortest distances
print("Shortest Distance Matrix:")

for i in range(V):
    for j in range(V):
        if dist[i][j] == INF:
            print("INF", end="\t")
        else:
            print(dist[i][j], end="\t")
    print()

# Display execution time
print(f"\nExecution Time: {end_time - start_time:.10f} seconds")


```
