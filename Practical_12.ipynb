```python
import time
from itertools import permutations

# Cost matrix representing distances between cities
graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

n = len(graph)
start_time = time.perf_counter()

min_cost = float('inf')
best_path = []

# Start from city 0
for path in permutations(range(1, n)):
    route = (0,) + path + (0,)
    cost = 0

    for i in range(len(route) - 1):
        cost += graph[route[i]][route[i + 1]]

    if cost < min_cost:
        min_cost = cost
        best_path = route

end_time = time.perf_counter()

print("Best Route:", " -> ".join(map(str, best_path)))
print("Minimum Travel Cost:", min_cost)

print("Execution Time:", end_time - start_time, "seconds")
```
