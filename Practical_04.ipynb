# Iterative Method

```python
import time

# Take user input
n = int(input("Enter a number: "))

# Start execution time
start_time = time.perf_counter()

# Iterative factorial
factorial = 1

for i in range(1, n + 1):
    factorial = factorial * i

# End execution time
end_time = time.perf_counter()

# Display result
print("\nFactorial using Iterative Method:", factorial)

# Display execution time
print(f"Execution Time: {end_time - start_time:.10f} seconds")

# Time and space complexity
print("\nTime Complexity: O(n)")
print("Space Complexity: O(1)")
```

# Recursive Method

```python
import time

# Recursive factorial function
def factorial_recursive(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial_recursive(n - 1)


# Take user input
n = int(input("Enter a number: "))

# Start execution time
start_time = time.perf_counter()

# Calculate factorial
factorial = factorial_recursive(n)

# End execution time
end_time = time.perf_counter()

# Display result
print("\nFactorial using Recursive Method:", factorial)

# Display execution time
print(f"Execution Time: {end_time - start_time:.10f} seconds")

# Time and space complexity
print("\nTime Complexity: O(n)")
print("Space Complexity: O(n)")
```
