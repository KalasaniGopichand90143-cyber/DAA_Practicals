# Linear search

```python
import time

# Take user input
n = int(input("Enter the number of elements: "))

arr = []
print("Enter the elements:")
for i in range(n):
    arr.append(int(input()))

# Take the element to search
target = int(input("\nEnter the element to search: "))

# Start execution time
start_time = time.perf_counter()

# Linear Search
position = -1

for i in range(n):
    if arr[i] == target:
        position = i
        break

# End execution time
end_time = time.perf_counter()

# Display result
if position != -1:
    print(f"\nElement {target} found at position {position + 1}")
else:
    print(f"\nElement {target} not found")

# Display execution time
print(f"Execution Time: {end_time - start_time:.10f} seconds")

# Display time complexity
print("\nTime Complexity:")
print("Best Case    : O(1)")
print("Average Case : O(n)")
print("Worst Case   : O(n)")
print("Space Complexity: O(1)")
```

# Binary Search

```python
import time

# Input from user
n = int(input("Enter the number of elements: "))

arr = []
print("Enter the elements:")
for i in range(n):
    arr.append(int(input()))

# Sort the array (Binary Search requires a sorted array)
arr.sort()

print("\nSorted Array:")
print(arr)

key = int(input("\nEnter the element to search: "))

# Start execution time
start_time = time.perf_counter()

# Binary Search
low = 0
high = n - 1
found = False
position = -1

while low <= high:
    mid = (low + high) // 2

    if arr[mid] == key:
        found = True
        position = mid
        break
    elif arr[mid] < key:
        low = mid + 1
    else:
        high = mid - 1

# End execution time
end_time = time.perf_counter()

# Calculate execution time
execution_time = (end_time - start_time) * 1_000_000  # Convert to microseconds

# Display result
if found:
    print(f"\nElement {key} found at position {position + 1}")
else:
    print(f"\nElement {key} not found in the array.")

# Display execution time
print(f"\nExecution Time: {execution_time:.2f} microseconds")

# Display time complexity
print("\nTime Complexity:")
print("Best Case    : O(1)")
print("Average Case : O(log n)")
print("Worst Case   : O(log n)")
```
