```python
import time

# Graph edges: (weight, vertex1, vertex2)
edges = [
    (2, 0, 1),
    (3, 1, 2),
    (6, 0, 3),
    (8, 1, 3),
    (5, 1, 4),
    (7, 2, 4),
    (9, 3, 4)
]

V = 5

# Start execution time
start_time = time.perf_counter()

# Sort edges by weight
edges.sort()

# Parent array
parent = list(range(V))

# Find operation
def find(v):
    if parent[v] != v:
        parent[v] = find(parent[v])
    return parent[v]

# Union operation
def union(u, v):
    root_u = find(u)
    root_v = find(v)

    if root_u != root_v:
        parent[root_v] = root_u
        return True

    return False

# Kruskal's Algorithm
mst = []
total_cost = 0

for weight, u, v in edges:
    if union(u, v):
        mst.append((u, v, weight))
        total_cost += weight

        if len(mst) == V - 1:
            break

# End execution time
end_time = time.perf_counter()

# Display result
print("Edges in Minimum Spanning Tree:")

for u, v, weight in mst:
    print(f"{u} - {v} : {weight}")

print("\nTotal Minimum Cost:", total_cost)
print(f"Execution Time: {end_time - start_time:.10f} seconds")

```
