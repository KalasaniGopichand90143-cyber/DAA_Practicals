```python
import time

# Graph represented using an adjacency matrix
graph = [
    [0, 2, 0, 6, 0],
    [2, 0, 3, 8, 5],
    [0, 3, 0, 0, 7],
    [6, 8, 0, 0, 9],
    [0, 5, 7, 9, 0]
]

V = len(graph)

# Start execution time
start_time = time.perf_counter()

selected = [False] * V
selected[0] = True

edges = 0
total_cost = 0

print("Edges in Minimum Spanning Tree:")

while edges < V - 1:
    minimum = float('inf')
    x = 0
    y = 0

    for i in range(V):
        if selected[i]:
            for j in range(V):
                if not selected[j] and graph[i][j] != 0:
                    if graph[i][j] < minimum:
                        minimum = graph[i][j]
                        x = i
                        y = j

    print(f"{x} - {y} : {graph[x][y]}")

    total_cost += graph[x][y]
    selected[y] = True
    edges += 1

# End execution time
end_time = time.perf_counter()

print("\nTotal Minimum Cost:", total_cost)
print(f"Execution Time: {end_time - start_time:.10f} seconds")
```
